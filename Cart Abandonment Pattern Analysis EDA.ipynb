```python
event_counts = {}

for chunk in pd.read_csv(file_path, chunksize=500000):
    counts = chunk['event_type'].value_counts()

    for event, count in counts.items():
        event_counts[event] = event_counts.get(event, 0) + count

event_counts
```

```python
unique_users = set()
unique_sessions = set()

for chunk in pd.read_csv(
    file_path,
    usecols=['user_id', 'user_session'],
    chunksize=500000
):
    unique_users.update(chunk['user_id'].dropna().unique())
    unique_sessions.update(chunk['user_session'].dropna().unique())

print("Unique users:", len(unique_users))
print("Unique sessions:", len(unique_sessions))
```

```python
del unique_users
del unique_sessions
```

```python
cart_sessions = set()
purchase_sessions = set()

for chunk in pd.read_csv(
    file_path,
    usecols=['user_session', 'event_type'],
    chunksize=500000
):
    cart_sessions.update(
        chunk.loc[chunk['event_type'] == 'cart', 'user_session'].unique()
    )

    purchase_sessions.update(
        chunk.loc[chunk['event_type'] == 'purchase', 'user_session'].unique()
    )

print("Sessions with cart:", len(cart_sessions))
print("Sessions with purchase:", len(purchase_sessions))
```

```python
cart_purchase_sessions = cart_sessions & purchase_sessions

print("Cart sessions:", len(cart_sessions))
print("Cart + purchase sessions:", len(cart_purchase_sessions))
print("Cart sessions without purchase:", len(cart_sessions - purchase_sessions))
```

```python
view_sessions = set()

for chunk in pd.read_csv(
    file_path,
    usecols=['user_session', 'event_type'],
    chunksize=500000
):
    view_sessions.update(
        chunk.loc[chunk['event_type'] == 'view', 'user_session'].unique()
    )

print("Sessions with view:", len(view_sessions))
```

```python
view_cart_sessions = view_sessions & cart_sessions

print("View sessions:", len(view_sessions))
print("View + cart sessions:", len(view_cart_sessions))
print("View sessions without cart:", len(view_sessions - cart_sessions))
```

```python
view_cart_purchase_sessions = view_sessions & cart_sessions & purchase_sessions

print("View + cart + purchase sessions:", len(view_cart_purchase_sessions))
```

```python
cart_file = r"C:\Users\DIVYA BASANTRAY\february_cart_events.csv"

first_chunk = True

for chunk in pd.read_csv(
    file_path,
    usecols=[
        'event_time',
        'event_type',
        'product_id',
        'category_id',
        'category_code',
        'brand',
        'price',
        'user_session'
    ],
    chunksize=500000
):
    
    cart_chunk = chunk[chunk['event_type'] == 'cart'].copy()

    if cart_chunk.empty:
        continue

    cart_chunk = cart_chunk.drop_duplicates()

    cart_chunk['purchased'] = cart_chunk['user_session'].isin(purchase_sessions)

    cart_chunk.to_csv(
        cart_file,
        mode='w' if first_chunk else 'a',
        index=False,
        header=first_chunk
    )

    first_chunk = False

print("Cart event file created.")
```

```python
category_cart = cart_df[
    ['user_session', 'category_code', 'purchased']
].copy()

category_cart['category_code'] = category_cart['category_code'].fillna('Unknown')
```

```python
category_cart = category_cart.drop_duplicates(
    ['user_session', 'category_code']
)
```

```python
category_summary = (
    category_cart
    .groupby('category_code')['purchased']
    .agg(
        cart_sessions='count',
        converted_sessions='sum'
    )
    .reset_index()
)

category_summary['abandoned_sessions'] = (
    category_summary['cart_sessions']
    - category_summary['converted_sessions']
)

category_summary['abandonment_rate'] = (
    category_summary['abandoned_sessions']
    / category_summary['cart_sessions']
    * 100
)

category_summary.sort_values(
    'cart_sessions',
    ascending=False
).head(15)
```

```python
category_summary_filtered = category_summary[
    category_summary['cart_sessions'] >= 1000
].sort_values(
    'abandonment_rate',
    ascending=False
)

category_summary_filtered.head(15)
```

```python
brand_cart = cart_df[
    ['user_session', 'brand', 'purchased']
].copy()

brand_cart['brand'] = brand_cart['brand'].fillna('Unknown')

brand_cart = brand_cart.drop_duplicates(
    ['user_session', 'brand']
)

brand_summary = (
    brand_cart
    .groupby('brand')['purchased']
    .agg(
        cart_sessions='count',
        converted_sessions='sum'
    )
    .reset_index()
)

brand_summary['abandoned_sessions'] = (
    brand_summary['cart_sessions']
    - brand_summary['converted_sessions']
)

brand_summary['abandonment_rate'] = (
    brand_summary['abandoned_sessions']
    / brand_summary['cart_sessions']
    * 100
)

brand_summary_filtered = brand_summary[
    brand_summary['cart_sessions'] >= 1000
].sort_values(
    'abandonment_rate',
    ascending=False
)

brand_summary_filtered.head(15)
```

```python
price_cart = cart_df[
    ['user_session', 'product_id', 'price', 'purchased']
].drop_duplicates(
    ['user_session', 'product_id']
)

price_cart['price_band'] = pd.cut(
    price_cart['price'],
    bins=[-1, 50, 100, 200, 350, 500, 1000, float('inf')],
    labels=[
        '$0–50',
        '$50–100',
        '$100–200',
        '$200–350',
        '$350–500',
        '$500–1,000',
        '$1,000+'
    ]
)

price_summary = (
    price_cart
    .groupby('price_band', observed=True)['purchased']
    .agg(
        cart_products='count',
        converted_products='sum'
    )
    .reset_index()
)

price_summary['abandoned_products'] = (
    price_summary['cart_products']
    - price_summary['converted_products']
)

price_summary['abandonment_rate'] = (
    price_summary['abandoned_products']
    / price_summary['cart_products']
    * 100
)

price_summary
```

```python
session_time = (
    cart_df
    .sort_values('event_time')
    .drop_duplicates('user_session')
    [['user_session', 'event_time', 'purchased']]
    .copy()
)

session_time['hour'] = session_time['event_time'].dt.hour

hour_summary = (
    session_time
    .groupby('hour')['purchased']
    .agg(
        cart_sessions='count',
        converted_sessions='sum'
    )
    .reset_index()
)

hour_summary['abandoned_sessions'] = (
    hour_summary['cart_sessions']
    - hour_summary['converted_sessions']
)

hour_summary['abandonment_rate'] = (
    hour_summary['abandoned_sessions']
    / hour_summary['cart_sessions']
    * 100
)

hour_summary
```

```python
session_products = (
    cart_df
    .groupby('user_session')
    .agg(
        unique_products=('product_id', 'nunique'),
        purchased=('purchased', 'first')
    )
    .reset_index()
)

session_products['product_count_band'] = pd.cut(
    session_products['unique_products'],
    bins=[0, 1, 2, 3, 5, 10, float('inf')],
    labels=[
        '1 product',
        '2 products',
        '3 products',
        '4–5 products',
        '6–10 products',
        '11+ products'
    ]
)

session_products['product_count_band'].value_counts().sort_index()
```

```python
session_summary = (
    session_products
    .groupby('product_count_band', observed=True)['purchased']
    .agg(
        cart_sessions='count',
        converted_sessions='sum'
    )
    .reset_index()
)

session_summary['abandoned_sessions'] = (
    session_summary['cart_sessions']
    - session_summary['converted_sessions']
)

session_summary['abandonment_rate'] = (
    session_summary['abandoned_sessions']
    / session_summary['cart_sessions']
    * 100
)

session_summary
```

```python
user_session_file = r"C:\Users\DIVYA BASANTRAY\february_user_sessions.csv"

first_chunk = True

for chunk in pd.read_csv(
    file_path,
    usecols=['user_id', 'user_session'],
    chunksize=500000
):
    
    chunk = chunk.drop_duplicates()

    chunk.to_csv(
        user_session_file,
        mode='w' if first_chunk else 'a',
        index=False,
        header=first_chunk
    )

    first_chunk = False

print("User-session file created.")
```

```python
user_session_df = pd.read_csv(user_session_file)

print("Rows:", len(user_session_df))
print("Unique users:", user_session_df['user_id'].nunique())
print("Unique sessions:", user_session_df['user_session'].nunique())
print("Duplicate user-session pairs:", user_session_df.duplicated().sum())
```

```python
user_session_df = user_session_df.drop_duplicates().copy()

print("Rows after deduplication:", len(user_session_df))
print("Duplicate pairs remaining:", user_session_df.duplicated().sum())
```

```python
cart_session_users = (
    user_session_df[
        user_session_df['user_session'].isin(
            cart_session_summary['user_session']
        )
    ]
    .drop_duplicates('user_session')
)

user_cart_sessions = (
    cart_session_users
    .groupby('user_id')['user_session']
    .nunique()
    .reset_index(name='cart_sessions')
)

user_cart_sessions['session_count_band'] = pd.cut(
    user_cart_sessions['cart_sessions'],
    bins=[0, 1, 2, 3, 5, float('inf')],
    labels=[
        '1 session',
        '2 sessions',
        '3 sessions',
        '4–5 sessions',
        '6+ sessions'
    ]
)

user_cart_sessions['session_count_band'].value_counts().sort_index()
```

```python
user_session_outcomes = (
    cart_session_users
    .merge(
        user_cart_sessions[['user_id', 'session_count_band']],
        on='user_id',
        how='left'
    )
    .merge(
        cart_session_summary,
        on='user_session',
        how='inner'
    )
)

user_summary = (
    user_session_outcomes
    .groupby('session_count_band', observed=True)['purchased']
    .agg(
        cart_sessions='count',
        converted_sessions='sum'
    )
    .reset_index()
)

user_summary['abandoned_sessions'] = (
    user_summary['cart_sessions']
    - user_summary['converted_sessions']
)

user_summary['abandonment_rate'] = (
    user_summary['abandoned_sessions']
    / user_summary['cart_sessions']
    * 100
)

user_summary
```
